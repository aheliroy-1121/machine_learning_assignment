```python
#2.2

students = [
    ["A", [90, 85, 80]],
    ["B", [70, 75, 60]],
    ["C", [65, 55, 50]]
]

highest = 0
topper = ""

for s in students:
  avg = sum(s[1])/len(s[1])
  print(s[0], "avg= ", avg)
  if avg > highest:
    highest = avg
    topper = s[0]

print("topper: ", topper  )
```

```python
departments = {
    "HR": ["Alice", "Bob"],
    "Engineering": ["Charlie", "David", "Eve"],
    "Sales": ["Frank", "Grace"]
}

print("Department and Employee Dictionary:")
print(departments)
```

```python
department_name = input("Enter a department name to view its employees: ")

if department_name in departments:
    print(f"Employees in {department_name}: {departments[department_name]}")
else:
    print(f"Department '{department_name}' not found.")
```

```python
print("\nEmployee count per department:")
for dept, employees in departments.items():
    print(f"{dept}: {len(employees)} employees")
```

```python
#8.1

set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8}

union = set1.union(set2)
print("union", union)
intersection = set1.intersection(set2)
print("intersection", intersection)
difference = set1.difference(set2)
print("difference", difference)
symmetric_difference = set1.symmetric_difference(set2)
print("symmetric_difference", symmetric_difference)
```
