create two 3 x 3 using data below with colum labels["c1, c2, c3"] and row levels ["r1, r2, r3"]
1. data frame addition
2. df subtraction
3. element wise multiplication
4. matix multiplication using dot

```python
import pandas as pd

A = pd.DataFrame(
    [[1,2,3],
     [4,5,6],
     [6,7,8]],
     index = ["r1", "r2", "r3"],
     columns = ["c1", "c2", "c3"]
)

B = pd.DataFrame(
    [[9,8,7],
     [6,5,4],
     [3,2,1]],
     index = ["r1", "r2", "r3"],
     columns = ["c1", "c2", "c3"]

)
print("data frame A:", A)
print("data frame B:", B)
print("data frame addition: ")
print(A + B)
print("\ndata frame subtraction: ")
print(A - B)
print("\nelement wise multiplication: ")
print(A * B)
print("\nmatrix multiplication using dot: ")
print(A.values.dot(B.values))
```

Student Maths Science English History
Amit 85 90 88 92
Bhavna 70 65 72 68
Chirag 95 91 93 96
Divya 60 75 70 80
Esha 88 84 86 90
Find:
1. Average marks of each student
2. Average marks of each subject
3. Highest and lowest marks
4. Student having the highest total marks
5. Sort the DataFrame by total marks in descending order

```python
import pandas as pd

data = {
    "student": ["Amit", "Bhavna", "Chirag", "Divya", "Esha"],
    "maths": [85, 70, 65, 97, 9],
    "science": [90, 65, 91, 75, 87],
    "bliology": [88, 72, 93, 80, 84],
    "history": [92, 68, 96, 87, 54]
}

df = pd.DataFrame(data)
df.set_index("student", inplace=True)
print("student datA:")
print(df)

df["avg"] = df.mean(axis = 1)

print("aVERAGE MARKS OF EACH STUDENT: ")
print(df["avg"])

print("avg marks of each sublect: ")
print(df.mean())

print("highest and lowest marks: ")
print("highest: ")
print(df.max())
print("lowest: ")
print(df.min())

df["total"] = df[["maths", "science", "bliology", "history"]].sum(axis=1)
print("student having the highest total marks: ")
print(df[df["total"] == df["total"].max()])

print("sort the DataFrame by total marks in descending order: ")
print(df.sort_values(by="total", ascending=False))
```

4. A dataset recording the temperature (°C) of a city over 7 days has some missing readings, represented below as a
DataFrame.

Day Day1 Day2 Day3 Day4 Day5 Day6 Day7

Temperature 30 NaN 32 29 NaN 35 31
Perform:
1. Identify missing values using isnull()
2. Fill missing values with the mean temperature
3. Fill missing values using forward fill (ffill)
4. Drop rows/columns containing missing values
5. Calculate the average temperature after cleaning the data


```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Day": ["Day1", "Day2", "Day3", "Day4", "Day5", "Day6", "Day7"],
    "Temperature": [30, np.nan, 32, 29, np.nan, 35, 31]
})

print("og df: ")
print(df)

print("\nmissing values: ")
print(df.isnull())

print("\n missing values with the mean temperature: ")
mean_temp = df["Temperature"].mean()
df_mean_filled = df.copy()
df_mean_filled["Temperature"] = df_mean_filled["Temperature"].fillna(mean_temp)
print(df_mean_filled)

print("\nmissing values using forward fill (ffill): ")
df_ffill = df.copy()
df_ffill["Temperature"] = df_ffill["Temperature"].ffill()
print(df_ffill)

print("\ndrop rows/columns containing missing values: ")
df_dropna = df.copy()
print(df_dropna)

print("\n avg temp after cleaning the data (after dropna): ")
avg_temp_after_dropna = df_dropna["Temperature"].mean()
print(avg_temp_after_dropna)
```

5. A company records the monthly sales of 4 products over 6 months in the DataFrame below.
Product Jan Feb Mar Apr May Jun
Product A 120 130 140 150 160 170
Product B 100 110 120 130 140 150
Product C 150 145 160 170 180 190
Product D 90 100 110 120 130 140
Perform the following:

a. Total sales of each product using groupby / sum()
b. Average monthly sales of each product
c. Best-selling product overall
d. Month with the highest overall sales
e. Standard deviation of sales for each product
e. Create a pivot table showing total sales per product per quarter (Q1: Jan–Mar, Q2: Apr–Jun)

```python
import pandas as pd

data = {
    "Product": ["A", "B", "C", "D"],
    "Jan": [120, 100, 150, 90],
    "Feb": [130, 110, 160, 90],
    "Mar": [140, 120, 170, 100],
    "Apr": [150, 130, 180, 120],
    "May": [160, 140, 190, 130],
    "Jun": [170, 150, 200, 150]
}
df = pd.DataFrame(data)
print("og df: ")
print(df)

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
df["Total"] = df[months].sum(axis=1)

print("\ntotal sales of each product: ")
print(df.groupby("Product")["Total"].sum())

print("\naverage monthly sales of each product: ")
print(df[months].mean())

print("\nbest-selling product overall: ")
print(df.loc[df["Total"].idxmax(), "Product"])

monthly_sales = df[months].values.flatten()
print("\nmonth with the highest overall sales: ")
print(pd.Series(monthly_sales).idxmax())

print("\nstandard deviation of sales for each product: ")
print(df[months].std())

df["q1"] = df[["Jan", "Feb", "Mar"]].sum(axis=1)
df["q2"] = df[["Apr", "May", "Jun"]].sum(axis=1)
pivot_table = pd.pivot_table(df, values=["q1", "q2"], index="Product")
print(pivot_table)
```
