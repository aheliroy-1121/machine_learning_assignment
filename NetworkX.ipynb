Question 5 — NetworkX
A school wants to visualize the friendship connections among a small group of students to understand
how they are socially linked.
Consider the following friendship connections:
Student A Student B
Alice Bob
Alice Charlie
Bob David
Charlie David
David Eva
Eva Frank
Charlie Frank
Task:
Using NetworkX, write a Python program to:
1. Create a graph representing the friendship network using the given connections.
2. Visualize the network graph with proper node labels, node colors, and edge styling.
3. Display the total number of nodes and edges in the graph.
4. Compute and print the degree of each student (number of friends).
5. Identify the student(s) with the highest number of friends.
6. Check and display whether a direct friendship exists between "Alice" and "Eva".
7. Find and display the shortest path (and its length) between "Alice" and "Frank".
8. Write 1–2 observations about the overall structure of the friendship network.
Expected Outcome:
A clearly labelled friendship network graph along with basic structural information such as node/edge
counts, degrees, and shortest paths between students.

```python
import matplotlib.pyplot as plt
import networkx as nx

# 1. Create a graph representing the friendship network using the given connections
G = nx.Graph()
connections = [
    ("Alice", "Bob"),
    ("Alice", "Charlie"),
    ("Bob", "David"),
    ("Charlie", "David"),
    ("David", "Eva"),
    ("Eva", "Frank"),
    ("Charlie", "Frank"),
]
G.add_edges_from(connections)

# 2. Visualize the network graph
plt.figure(figsize=(8, 6))
pos = nx.spring_layout(G, seed=42)  # For consistent layout
nx.draw(
    G,
    pos,
    with_labels=True,
    node_color="skyblue",
    node_size=2000,
    edge_color="darkred",
    width=2,
    font_size=12,
    font_weight="bold",
)
plt.title("Question 5: School Friendship Network", fontsize=14)
plt.show()

# 3. Display the total number of nodes and edges
print("--- Question 5 Results ---")
print(f"Total number of students (nodes): {G.number_of_nodes()}")
print(f"Total number of friendships (edges): {G.number_of_edges()}")

# 4. Compute and print the degree of each student
degrees = dict(G.degree())
print("\nDegree of each student (number of friends):")
for student, deg in degrees.items():
    print(f"- {student}: {deg}")

# 5. Identify the student(s) with the highest number of friends
max_degree = max(degrees.values())
top_students = [s for s, d in degrees.items() if d == max_degree]
print(
    f"\nStudent(s) with the highest number of friends: {', '.join(top_students)} with {max_degree} friends"
)

# 6. Check and display whether a direct friendship exists between 'Alice' and 'Eva'
direct_friendship = G.has_edge("Alice", "Eva")
print(
    f"\nDirect friendship between Alice and Eva exists? {'Yes' if direct_friendship else 'No'}"
)

# 7. Find and display the shortest path (and its length) between 'Alice' and 'Frank'
shortest_path = nx.shortest_path(G, source="Alice", target="Frank")
path_length = nx.shortest_path_length(G, source="Alice", target="Frank")
print(f"\nShortest path between Alice and Frank: {shortest_path}")
print(f"Length of the shortest path: {path_length}")

# 8. Observations
print(
    "\nObservations:\n"
    "1. Charlie and David act as central hubs or bridges in this social circle, possessing the highest degree connections.\n"
    "2. The network has a relatively small diameter, meaning information or rumors can travel quickly between any two students (e.g., Alice to Frank takes only 2 steps)."
)
```

Question 6 — NetworkX
A telecom company wants to analyze its network of towers to understand connectivity and identify
important hubs.
Consider the following tower connections (edges represent direct signal links):
Tower A Tower B
T1 T2
T1 T3
T2 T4
T3 T4
T4 T5
T5 T6
T3 T6
T6 T7
T2 T7
Task:
Using NetworkX, write a Python program to:
1. Create a graph representing the tower network using the given edges.
2. Visualize the network graph with proper node labels, node colors, and edge styling.
3. Calculate and display the total number of nodes and edges.
4. Compute the degree of each node and identify the tower(s) with the highest degree (most
connections).
5. Calculate and display the shortest path (and its length) between T1 and T7.
6. Compute and display the following centrality measures for all nodes: Degree Centrality,
Betweenness Centrality, and Closeness Centrality.
7. Identify whether the network is fully connected. If not, identify the number of connected components.
8. Highlight the tower with the highest betweenness centrality on the graph visualization (e.g., using a
different color/size).
9. Write 2–3 observations about which towers are most critical to the network based on the computed
metrics.
Expected Outcome:
A visualized network graph along with computed structural metrics that help identify the most important
and best-connected towers in the network.

```python
import matplotlib.pyplot as plt
import networkx as nx

# 1. Create a graph representing the tower network using the given edges
G_tel = nx.Graph()
tower_edges = [
    ("T1", "T2"),
    ("T1", "T3"),
    ("T2", "T4"),
    ("T3", "T4"),
    ("T4", "T5"),
    ("T5", "T6"),
    ("T3", "T6"),
    ("T6", "T7"),
    ("T2", "T7"),
]
G_tel.add_edges_from(tower_edges)

# 3. Calculate and display the total number of nodes and edges
print("--- Question 6 Results ---")
print(f"Total number of towers (nodes): {G_tel.number_of_nodes()}")
print(f"Total number of signal links (edges): {G_tel.number_of_edges()}")

# 4. Compute the degree of each node and identify the tower(s) with the highest degree
degrees_tel = dict(G_tel.degree())
max_deg_tel = max(degrees_tel.values())
top_towers = [t for t, d in degrees_tel.items() if d == max_deg_tel]
print("\nTower Degrees:")
for tower, deg in degrees_tel.items():
    print(f"- {tower}: {deg}")
print(
    f"Tower(s) with the highest connections: {', '.join(top_towers)} ({max_deg_tel} connections)"
)

# 5. Calculate and display the shortest path (and its length) between T1 and T7
sp_tel = nx.shortest_path(G_tel, source="T1", target="T7")
sp_len_tel = nx.shortest_path_length(G_tel, source="T1", target="T7")
print(f"\nShortest path between T1 and T7: {sp_tel}")
print(f"Length of shortest path: {sp_len_tel}")

# 6. Compute centrality measures
deg_cent = nx.degree_centrality(G_tel)
bet_cent = nx.betweenness_centrality(G_tel)
clo_cent = nx.closeness_centrality(G_tel)

print("\nCentrality Measures:")
for node in G_tel.nodes():
    print(
        f"Tower {node} -> Degree: {deg_cent[node]:.2f}, Betweenness: {bet_cent[node]:.2f}, Closeness: {clo_cent[node]:.2f}"
    )

# 7. Check connectivity
is_connected = nx.is_connected(G_tel)
num_components = nx.number_connected_components(G_tel)
print(f"\nIs the network fully connected? {is_connected}")
print(f"Number of connected components: {num_components}")

# 8. Highlight the tower with the highest betweenness centrality on the graph visualization
max_bet_tower = max(bet_cent, key=bet_cent.get)
print(
    f"\nTower with the highest betweenness centrality (critical bottleneck): {max_bet_tower}"
)

node_colors = [
    "orange" if node == max_bet_tower else "lightgreen"
    for node in G_tel.nodes()
]
node_sizes = [
    3000 if node == max_bet_tower else 1500 for node in G_tel.nodes()
]

plt.figure(figsize=(8, 6))
pos_tel = nx.spring_layout(G_tel, seed=42)
nx.draw(
    G_tel,
    pos_tel,
    with_labels=True,
    node_color=node_colors,
    node_size=node_sizes,
    edge_color="grey",
    width=2,
    font_size=12,
    font_weight="bold",
)
plt.title(
    f"Question 6: Telecom Tower Network\n(Highlighted: {max_bet_tower} with highest betweenness)",
    fontsize=12,
)
plt.show()

# 9. Observations
print(
    "\nObservations:\n"
    "1. Towers like T3 and T4 have high degrees and serve as major localized connectivity hubs within the grid.\n"
    "2. The tower with the highest betweenness centrality acts as a critical bridge for routing signals across different segments of the network; its failure would heavily disrupt communication paths."
)
```
